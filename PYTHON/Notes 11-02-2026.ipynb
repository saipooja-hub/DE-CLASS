```python
#LOOPS:
     Loops are used to execute a block of code repeatedly,There are 2 main types of loops:1.for loop 2.while loop.
FOR LOOP:
    For loops, used for iterating over sequences with a known number of iterations.A for loop iterates over the items of any iterable object, such as a list, tuple, dictionary, string, or range.
WHILE LOOP:
    A while loop continues executing a block of code as long as its condition remains true. This is useful when the number of iterations is not known in advance and depends on a dynamic condition.
```

```python
#While Loop:
This loop continues until the condition is True.
```

```python
n = 9

# initialize sum and counter
sum = 0
i = 1

while i <= n:
    sum = sum + i
    i = i+1    # update counter

# print the sum
print("The sum is", sum)
```

```python
#Break and continue:
#BREAK:
    The break statement immediately terminates the entire loop (both for and while loops) in which it is used, and the program's control flow moves to the next statement after the loop
```

```python
# Use of break statement inside loop

for val in "string":
    if val == "r":
        break
    print(val)
print("The end")
```

```python
#Continue:
        The continue statement skips the remaining code inside the loop for the current iteration only, and the control flow returns to the start of the loop to begin the next iteration.
```

```python
# Program to show the use of continue statement inside loops

for val in "string":
    if val == "r":
        continue
    print(val)

print("The end")
```
