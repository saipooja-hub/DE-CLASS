```python
#partition:
        partition() when you need to split a string into exactly two parts and you want to keep the separator itself, such as parsing key-value pairs or file paths where you expect a clear delimiter.
#filename = "document.txt"
name, sep, extension = filename.partition(".")
# Output: name='document', sep='.', extension='txt'
```

```python
#Nesting list:
            list within a list is  called nesting list.a list that contains other lists as its elements. This structure is commonly used to represent multi-dimensional data like matrices or tables. 
```

```python
lst_1=[1,2,3]
lst_2=[4,5,6]
lst_3=[7,8,9]

#make all lists to form a matrix
matrix = [lst_1,lst_2,lst_3]
```

```python
matrix
```

```python
matrix[0][1]
```

```python
matrix[1][0]
```

```python
matrix[2][2]
```

```python
#list comprehension:
                List comprehension offers a shorter syntax when you want to create a new list based on the values of an existing list.
example:
        Based on a list of fruits, you want a new list, containing only the fruits with the letter "a" in the name.
```

```python
first_col = [row[0] for row in matrix]

first_col
```

```python
first_col = []
for row in matrix:
    first_col.append(row[0])
print(first_col)
```
