```python
#Generators:
            Generators are a special, more convenient type of iterator that can be created using generator functions or generator expressions
Generator Functions are defined like normal functions, but use the yield statement instead of return. Each yield pauses the function and returns a value, and the function's local state is saved. When iteration resumes, execution continues from where it left off.
Generator Expressions are a more compact syntax similar to list comprehensions, using parentheses () instead of square brackets []
# Generator function
def count_generator(limit):
    for i in range(limit):
        yield i

# Using the generator
gen = count_generator(3)
for num in gen:
    print(num) 
# Output:
# 0
# 1
# 2
```

```python
def genfibon(n):
    '''
    Generate a fibonacci sequence up to n
    '''
    a = 1
    b = 1
    for i in range(n):
        yield a
        a,b = b,a+b
```

```python
genfibon(5)
```

```python
for num in genfibon(12):
    print(num)
```

```python
def fibon(n):
    a = 1
    b = 1
    output = []
    
    for i in range(n):
        output.append(a)
        a,b = b,a+b
        
    return output
```

```python
fibon(10)
```

```python
#map function:
            The map() function applies a given function to each item in an iterable and returns a new iterable (a map object) with the results
#squaring all numbers in a list:
items = [1, 2, 3, 4, 5]
# Use a lambda function to square each item
squared = list(map(lambda x: x**2, items))
# Output: [1, 4, 9, 16, 25]
```

```python
#Reduce:
        The reduce() function applies a rolling computation to sequential pairs of values in an iterable, ultimately reducing it to a single value
It is not a built-in function like map and filter; it must be imported from the functools module.
# Calculating the product of all numbers in a list. The function passed to reduce must take two arguments.
from functools import reduce

list_of_numbers = [1, 2, 3, 4]
# Use a lambda function to multiply two numbers cumulatively
product = reduce((lambda x, y: x * y), list_of_numbers)
# Output: 24
```

```python
#Find the maximum of a sequence (This already exists as max())
max_find = lambda a,b: a if (a > b) else b
lst =[47,49,42,55,56]
```

```python
#Filter:
        The filter() function constructs a new iterable from elements of the original iterable for which a given function returns True.
#Extracting even numbers from a list. The function passed to filter must return a boolean value.
number_list = [1, 2, 3, 4, 5, 6]
# Use a lambda function to check for even numbers
even_numbers = list(filter(lambda x: x % 2 == 0, number_list))
# Output: [2, 4, 6]
```
