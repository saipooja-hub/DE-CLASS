```python
#there are 2 important concepts in copy those are:
1.shallow copy
2.deep copy
```

```python
#shallow copy: the changes made to a mutable nested object in the copy will affect the original, and vice versa.

import copy

original = [1,2,3,5] 

shallow = copy.copy(original)

shallow[0] = 12

print('original', original)
print('shallow', shallow)
```

```python
#deep copy it creates the copy of the original value.recursively duplicates all nested objects within it. This ensures that the original and the copy are fully independent, and changes to one do not affect the other.

import copy

original = [[1,2],[3,4]] 

deep = copy.deepcopy(original)

deep[0][0] = 12

print('original', original)
print('Deep', deep)
```

```python
def is_prime(num):
    '''
    Naive method of checking for primes. 
    '''
    for n in range(2,num):
        if num % n == 0:
            print('not prime')
            break
    else: # If never mod zero, then prime
        print('prime')
```

```python
is_prime(9)
```

```python
is_prime(8)
```
